```python
cafe_sales_df = pd.read_csv('C:\\Users\\user\\Desktop\\aiml\\Daily-Practice\\Day 20\\dirty_cafe_sales.csv')

cafe_sales_df.head()
```

```python
import numpy as np

cafe_sales_df['Price Per Unit'] = cafe_sales_df['Price Per Unit'].replace(['UNKNOWN','ERROR'], np.nan)
cafe_sales_df['Price Per Unit'] = pd.to_numeric(cafe_sales_df['Price Per Unit'])
```

```python
cafe_sales_df['Price Per Unit'] = cafe_sales_df.groupby('Item')['Price Per Unit'].transform('first')

cafe_sales_df['Price Per Unit']
```

```python
cafe_sales_df['Quantity'].value_counts()
```

```python
mode_quantity = cafe_sales_df['Quantity'].mode()[0]
mode_quantity
```

```python
cafe_sales_df['Quantity'] = cafe_sales_df['Quantity'].replace(['ERROR','UNKNOWN'], mode_quantity) 
```
