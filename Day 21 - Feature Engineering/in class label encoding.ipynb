### Encoding ###

```python
# 1. Manual way
# label encoding 

# - takes a class, assign new integer value to every new class
# column that you can encode 

penguins['species'].unique()


species_codes = {
    'Adelie' : 0,
    'Chinstrap': 1,
    'Gentoo': 2
}

#attack column               
penguins['species_encoded'] = penguins['species'].map(species_codes) # map 

penguins['species_encoded']
```

```python
penguins['island'].unique()
```

```python
island_codes = {
    'Torgersen':0,
    'Biscoe':1,
    'Dream':2
}

penguins['island_encodes'] = penguins['island'].map(island_codes)
penguins['island_encodes']
```

```python
# 2. Factorization
# factorize (another way)
# species still contains the species name 

#factorize will take the column and assign numbers to the new groups 
species_factorized = pd.factorize(penguins['species']) 
# print(species_factorized) # ada 2D, 1st result. 2 nama index
# new column                   #
penguins['species_factorized'] = species_factorized[0] 
penguins['species_factorized']
```
