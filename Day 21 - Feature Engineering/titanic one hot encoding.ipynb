***One hot encoding***

```python
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
# using pandas - one hot encoder
titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
#get dummies method - categories --> Columns
titanic_onehot=pd.get_dummies(titanic, columns=['sex']) # get dummies automatically ignores any missing values
titanic_onehot
```

```python
# using sklearn onehot encoder
from sklearn.preprocessing import OneHotEncoder
import numpy as np
```

pl, pw, sl, sw, species, setosa, virginica, versicolor

Species --> object
drop the column betfore fitting into the model 

```python
titanic_onehot=titanic.copy()

#create the onehotencoder  # returns nump array  #
encoder = OneHotEncoder(sparse_output=False,handle_unknown='ignore') 

categorical_cols = ['embark_town','alive']

encoded_values = encoder.fit_transform(titanic_onehot[categorical_cols])

#name of new column that dont exist
new_col = encoder.get_feature_names_out(categorical_cols)
new_col


titanic_onehot[new_col] = encoded_values
titanic_onehot

titanic_onehot.drop(columns=['embark_town'],inplace=True)
titanic_onehot
```

```python
# 4.using sklearn labelencoder method
from sklearn.preprocessing import LabelEncoder

# Create LabelEncoder distance
encoder = LabelEncoder()

titanic['alive'] = encoder.fit_transform(titanic['alive'])
titanic
```
