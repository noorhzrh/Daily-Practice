```python
iris_onehot=iris.copy()

#create the onehotencoder  # returns nump array  #
encoder = OneHotEncoder(sparse_output=False,handle_unknown='ignore') 

categorical_cols = ['species']

iris_onehot = encoder.fit_transform(iris_onehot[categorical_cols])
```

```python
penguins_onehot=penguins.copy()

#create the onehotencoder  # returns nump array  #
encoder = OneHotEncoder(sparse_output=False,handle_unknown='ignore') 

categorical_cols = ['species']

penguins_onehot = encoder.fit_transform(penguins_onehot[categorical_cols])
penguins_onehot
```

```python
iris_onehot=iris.copy()

#create the onehotencoder  # returns nump array  #
encoder = OneHotEncoder(sparse_output=False,handle_unknown='ignore') 

categorical_cols = ['species']

encoded_values = encoder.fit_transform(iris_onehot[categorical_cols])

#name of new column that dont exist
new_col = encoder.get_feature_names_out(categorical_cols)
new_col
#'species_setosa', 'species_versicolor', 'species_virginica'],


iris_onehot[new_col] = encoded_values
iris_onehot

iris_onehot.drop(columns=['species'],inplace=True)
iris_onehot
```
