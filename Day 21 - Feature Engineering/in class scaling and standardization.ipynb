##scaling

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
data = pd.DataFrame({
    'age': [20, 50, 100],
    'income': [20000,50000,100000]
})

# upscale/downscale
scaler = MinMaxScaler()
scaler # object

# Normalization you want to check what the data will look like in a scale of 0 to 1.

normalize_data = scaler.fit_transform(data)

normalize_data # largest become 1, smallest become 0
```

```python
df_normalized = pd.DataFrame(normalize_data, columns=['age','income'])
df_normalized
```

- Why do we normalize data?

- easier to calculate

a lot prefer to salary to age in the given example

- our models only use numbers
