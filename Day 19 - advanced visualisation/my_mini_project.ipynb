```python
#change data type of date from string
crimedf['date'] = pd.to_datetime(crimedf['date'])
#extract the year
crimedf['date'] = crimedf['date'].dt.year
crimedf['date'] = crimedf['date'].astype(int)
```

```python
crimedf.info()
```

```python
# filter Malaysia and years 2016–2023
malaysia_df = crimedf[
    (crimedf['state'] == 'Malaysia') &
    (crimedf['district'] == 'All') &
    (crimedf['date'].between(2016, 2023)) &
    (crimedf['category'].isin(['assault', 'property']))
]

malaysia_df
```

```python
crime_counts = (malaysia_df.groupby(['category','date'])['crimes'].sum().reset_index())
# groupby date and category 
```

```python
fig = px.bar(
    crime_counts,
    x='date',
    y='crimes',
    color='category',
    barmode='group',
    title='Crime Counts in Malaysia (2013–2016): Assault vs Property'
)

fig.show()
```
