```python
#change data type of date from string
crimedf['date'] = pd.to_datetime(crimedf['date'])
#extract the year
crimedf['date'] = crimedf['date'].dt.year
crimedf['date'] = crimedf['date'].astype(int)
```

```python
crimedf.info()
```

```python
# filter Malaysia and years 2016–2023
malaysia_df = crimedf[
    (crimedf['state'] == 'Malaysia') &
    (crimedf['district'] == 'All') &
    (crimedf['date'].between(2016, 2023)) &
    (crimedf['category'].isin(['assault', 'property']))
]

malaysia_df.head(50)
```

```python
crime_countsdf = (malaysia_df.groupby(['category','date'])['crimes'].sum().reset_index())
crime_countsdf
# crimes are groupby date and category 
```

```python

bargraph = px.bar(crime_countsdf, x='date', y='crimes', color='category', title='Crime Counts in Malaysia (2016–2023): Assault vs Property')    # barmode='group'
# pakai crime_countsdf, 
bargraph.show()
```

```python
bystate_df = crimedf[
    (crimedf['district'] == 'All') &
    (crimedf['state'] != 'Malaysia') &
    (crimedf['type'] == 'all') &
    (crimedf['category'].isin(['assault','property']))
]

bystate_df
```

```python
crimebystate_countsdf = (bystate_df.groupby(['state'])['crimes'].sum().reset_index())
crimebystate_countsdf
```

```python
fig = px.bar(
    crimebystate_countsdf, x='state', y='crimes', color='state', title='Crime Counts in Malaysia (2016–2023) by states')    # barmode='group'
fig.show()
```
