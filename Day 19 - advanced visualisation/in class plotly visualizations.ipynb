```python
import matplotlib.pyplot as plt   
import matplotlib as mpl
import pandas as pd
import seaborn as sns  
```

```python
import plotly.express as px 
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
# rc parameters
# runtime configurations
plt.rcParams["figure.figsize"] = (10,7)
plt.rcParams['lines.linewidth'] = 5
plt.rcParams['lines.linestyle'] = "--"
plt.rcParams['grid.linestyle'] = "--"
```

```python
mpl.rcParams.update({
    "figure.figsize": (8, 4),
    "figure.dpi": 120,
    "axes.grid": True,
    "grid.alpha": 0.3,
    "lines.linewidth": 2,
    "lines.linestyle": '-',
    "font.size": 12,
    "axes.titlesize": 14,
    "axes.labelsize": 12,
    "legend.frameon": False,
})
```

```python
plt.scatter(x=tips['total_bill'], y=tips['tip'])
```

```python
# dpi will define the pixel density
sns.set_theme(
    style='whitegrid',
    context= 'talk',
    rc={
        'figure.figsize': (7,4),
        'figure.dpi': 120,
        'axes.titlesize': 8,
        'axes.labelsize': 10,
        'lines.linewidth': 2,
        'lines.linestyle': '--',
        'legend.frameon':False,
        'xtick.labelsize': 7,
        'ytick.labelsize': 7
    }
)
```

```python
sns.set(style='darkgrid')
sns.pairplot(tips, hue='time')
```

```python
#join plot

sns.jointplot(tips, x='total_bill', y='tip', kind='kde', fill = True )
```

```python
sns.boxenplot(tips, x = 'day' , y = 'total_bill')
```

```python
sns.countplot(tips, x = 'day', hue= "time")
```

```python
# plotly 
import plotly.express as px

barplot = px.bar(tips, x= 'day' , y= 'total_bill') 

barplot.show()
```

```python
# how will you plot mean for everyday
# for eachj dayyyyyy

bill_by_day = tips.groupby('day', as_index=False)['total_bill'].mean()
bill_by_day
```

```python
# plotly 
import plotly.express as px

barplot = px.bar( bill_by_day, x= 'day' , y= 'total_bill') 

barplot.show()

# plot the mean of total bill on each day
```

```python
scatterbilltip = px.scatter(
    tips,
    x= 'total_bill',
    y= 'tip',
    size=  'size',
    color= 'day',
    hover_data= ['sex','smoker','time']

)

scatterbilltip.show()
```
