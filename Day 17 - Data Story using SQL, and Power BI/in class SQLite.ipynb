```python
# connect to a database
# there is no database, when we create a connection, automatically create it

#variable = sql.connect (create db)
connection = sql.connect('test.db')

# we need a cursor object: to point to records in the database - move through the db

cursor = connection.cursor()
```

```python
cursor.execute( '''              
CREATE TABLE students(
    id INT PRIMARY KEY,
    name VARCHAR(10),
    city VARCHAR(10)
)
''')
```

```python
# commit the changes and close the connection
connection.commit()
```

```python
cursor.execute("INSERT INTO students (id,name,city) VALUES (5,'Hazirah','Seremban')")
cursor.execute("INSERT INTO students (id,name,city) VALUES (2,'Hazirah','Seremban')")
cursor.execute("INSERT INTO students (id,name,city) VALUES (3,'Hazirah','Seremban')")
cursor.execute("INSERT INTO students (id,name,city) VALUES (4,'Hazirah','Seremban')")
```
